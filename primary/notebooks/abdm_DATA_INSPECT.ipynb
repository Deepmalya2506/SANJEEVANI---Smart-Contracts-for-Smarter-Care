Clean and spatially group the hospital directory
This pipeline keeps a useful directory table, treats `0` as missing where the source uses it as a placeholder, and preserves hospitals without coordinates for later enrichment.


```python
# Load as strings first so phone numbers, pincodes, and IDs keep their formatting.
raw_hospitals = pd.read_csv(data_path, dtype="string")

# The source uses 0 and blank strings for unavailable values.
placeholder_values = {"0": pd.NA, "": pd.NA, "nan": pd.NA, "None": pd.NA}
hospitals = raw_hospitals.replace(placeholder_values)

# Keep fields that support identity, discovery, matching, capacity, and contact.
useful_columns = [
    "Sr_No", "Hospital_Name", "Hospital_Category", "Hospital_Care_Type",
    "Discipline_Systems_of_Medicine", "Address_Original_First_Line", "Location",
    "State", "District", "Subdistrict", "Pincode", "Telephone", "Mobile_Number",
    "Emergency_Num", "Ambulance_Phone_No", "Website", "Specialties", "Facilities",
    "Total_Num_Beds", "Establised_Year", "Emergency_Services", "Location_Coordinates",
]
useful_columns = [column for column in useful_columns if column in hospitals.columns]
hospitals = hospitals[useful_columns].copy()

# Parse the source "latitude, longitude" field into numeric spatial columns.
coordinates = hospitals["Location_Coordinates"].str.split(",", n=1, expand=True)
hospitals["latitude"] = pd.to_numeric(coordinates[0].str.strip(), errors="coerce")
hospitals["longitude"] = pd.to_numeric(coordinates[1].str.strip(), errors="coerce")
hospitals = hospitals.drop(columns="Location_Coordinates")

# Normalize text without changing phone numbers or identifier columns.
text_columns = hospitals.select_dtypes(include="string").columns
hospitals[text_columns] = hospitals[text_columns].apply(lambda column: column.str.strip())

# This directory is India-focused; reject impossible and out-of-country coordinates.
valid_coordinates = (
    hospitals["latitude"].between(6, 38.5)
    & hospitals["longitude"].between(68, 98.5)
)
hospitals.loc[~valid_coordinates, ["latitude", "longitude"]] = pd.NA

# Convert capacity to numeric and discard implausible source values.
raw_beds = pd.to_numeric(hospitals["Total_Num_Beds"], errors="coerce")
invalid_bed_count = int((raw_beds > 10000).sum())
hospitals["total_beds"] = raw_beds.where(raw_beds.between(0, 10000))
hospitals = hospitals.drop(columns="Total_Num_Beds")

# A rounded key groups nearby facilities while preserving the original coordinates.
hospitals["position_group"] = (
    hospitals["latitude"].round(3).astype("string")
    + ", "
    + hospitals["longitude"].round(3).astype("string")
)
hospitals.loc[hospitals[["latitude", "longitude"]].isna().any(axis=1), "position_group"] = pd.NA

spatial_hospitals = hospitals.dropna(subset=["latitude", "longitude"]).copy()
position_counts = spatial_hospitals["position_group"].value_counts()
spatial_hospitals["hospitals_at_position"] = spatial_hospitals["position_group"].map(position_counts)

# One row per rounded position, useful for map markers and candidate search.
grouped_by_position = (
    spatial_hospitals.groupby("position_group", as_index=False)
    .agg(
        latitude=("latitude", "mean"),
        longitude=("longitude", "mean"),
        hospital_count=("Hospital_Name", "size"),
        hospitals=("Hospital_Name", lambda values: " | ".join(values.dropna().unique()[:10])),
        states=("State", lambda values: " | ".join(values.dropna().unique())),
        districts=("District", lambda values: " | ".join(values.dropna().unique())),
    )
    .sort_values(["hospital_count", "position_group"], ascending=[False, True])
)

# Useful quality checks for the next data-enrichment step.
data_quality = pd.DataFrame({
    "metric": [
        "total source rows", "rows with valid India coordinates", "rows without usable coordinates",
        "unique rounded positions", "positions shared by 2+ hospitals", "implausible bed values removed",
    ],
    "value": [
        len(hospitals), len(spatial_hospitals), len(hospitals) - len(spatial_hospitals),
        len(grouped_by_position), int((position_counts > 1).sum()), invalid_bed_count,
    ],
})

display(data_quality)
display(grouped_by_position.head(20))
```

```python
# Remove only columns that are completely empty after placeholder normalization.
empty_columns = hospitals.columns[hospitals.isna().all()].tolist()
processed_hospitals = hospitals.drop(columns=empty_columns)

state_summary = (
    processed_hospitals.groupby("State", dropna=False)
    .agg(
        hospital_count=("Hospital_Name", "size"),
        mapped_count=("position_group", lambda values: values.notna().sum()),
        total_beds=("total_beds", "sum"),
        districts=("District", "nunique"),
    )
    .assign(mapped_percent=lambda table: (100 * table["mapped_count"] / table["hospital_count"]).round(1))
    .sort_values("hospital_count", ascending=False)
    .reset_index()
)

# These files are ready for later PostGIS/H3 loading and dashboard prototyping.
processed_hospitals.to_csv("processed_hospitals.csv", index=False)
grouped_by_position.to_csv("hospitals_by_position.csv", index=False)
state_summary.to_csv("hospital_state_summary.csv", index=False)

display(state_summary.head(15))
print(f"Dropped completely empty columns: {empty_columns}")
print("Saved: processed_hospitals.csv, hospitals_by_position.csv, hospital_state_summary.csv")
```

```python
imp_df.dropna(subset=['Hospital_Name'], inplace=True)
```

```python
# Strip whitespace:
imp_df = imp_df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)

# Standardize case (title case works well for Indian addresses):
imp_df['Address_Original_First_Line'] = imp_df['Address_Original_First_Line'].str.title()

# Replace nulls with empty strings:
imp_df = imp_df.fillna({'Address_Original_First_Line': '', 'Location': '', 'District': '', 'State': '', 'Pincode': ''})
```

```python
def smart_add(base, extra):
    if extra.lower() in base.lower():
        return base
    return f"{base}, {extra}" if extra else base
```

```python
imp_df['Address_Clean'] = imp_df.apply(
    lambda row: smart_add(
        smart_add(
            smart_add(
                row['Address_Original_First_Line'], row['Location']
            ), row['District']
        ), row['State']
    ),
    axis=1
)
```

```python
imp_df['Address_Clean'] = imp_df.apply(
    lambda row: f"{row['Address_Clean']} - {row['Pincode']}" if row['Pincode'] else row['Address_Clean'],
    axis=1
)
```

```python
imp_df['Full_Address'] = imp_df.apply(
    lambda r: {
        "address_line": r['Address_Original_First_Line'],
        "location": r['Location'],
        "district": r['District'],
        "state": r['State'],
        "pincode": r['Pincode']
    },
    axis=1
)
```

```python
import pandas as pd
import time
from geopy.geocoders import Nominatim
from tqdm import tqdm

geolocator = Nominatim(user_agent="sanjeevani_app_v1", timeout=10)

# 1. Identify indices where lat/long are missing
missing_mask = imp_df['latitude'].isna() | imp_df['longitude'].isna()
missing_indices = imp_df[missing_mask].index

# 2. Select a subset (e.g., first 50 missing rows -> takes ~50 seconds)
SUBSET_SIZE = 200
subset_indices = missing_indices[:SUBSET_SIZE]

print(f"Processing {len(subset_indices)} rows (~{len(subset_indices)} seconds total)...")

# 3. Geocode and update in place with a progress bar
for idx in tqdm(subset_indices):
    address = imp_df.loc[idx, 'Address_Clean']
    
    if pd.isna(address) or not str(address).strip():
        continue
        
    time.sleep(1) # Respect OSM rate limit
    
    try:
        location = geolocator.geocode(str(address))
        if location:
            imp_df.loc[idx, 'latitude'] = location.latitude
            imp_df.loc[idx, 'longitude'] = location.longitude
    except Exception:
        continue
```

```python
# 1. Filter valid lat/long rows
extracted_df = imp_df.dropna(subset=['latitude', 'longitude']).copy()

# 2. Generate unique HFR Mock IDs (e.g., HFR_MOCK_000001, HFR_MOCK_000002)
extracted_df['mvp_hfr_id'] = [f"HFR_MOCK_{i+1:06d}" for i in range(len(extracted_df))]

# 3. Add default metadata columns expected by Supabase
extracted_df['source_record_id'] = extracted_df.index.astype(str)
extracted_df['official_hfr_id'] = None
extracted_df['source'] = 'data.gov.in'

# 4. Rename columns to match Supabase table exactly (lowercase snake_case)
final_df = extracted_df.rename(columns={
    'Hospital_Name': 'hospital_name',
    'Address_Clean': 'address'
})

# 5. Select only the matching schema columns
supabase_columns = [
    'mvp_hfr_id',
    'official_hfr_id',
    'hospital_name',
    'source_record_id',
    'address',
    'latitude',
    'longitude',
    'source'
]

# 6. Save to CSV
final_df[supabase_columns].to_csv('../data/processed/geocoded_output.csv', index=False)
```
